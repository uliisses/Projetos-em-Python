```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configurar estilo dos gráficos
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
tabela = pd.read_csv('cancelamentos.csv')
tabela = tabela.drop('CustomerID', axis=1)
display(tabela.head())
```

```python
# Identificando e removendo valores vazios
tabela = tabela.dropna()
print(f"Dados após limpeza: {len(tabela)} registros")
```

```python
# Visualizar distribuição de cancelamentos
fig, ax = plt.subplots(figsize=(10, 6))
cancelamento_counts = tabela['cancelou'].value_counts()
cores = ['#2ecc71', '#e74c3c']
cancelamento_counts.plot(kind='bar', ax=ax, color=cores, edgecolor='black', linewidth=1.5)
ax.set_title('Distribuição de Cancelamentos de Clientes', fontsize=14, fontweight='bold')
ax.set_xlabel('Cancelou (0=Não, 1=Sim)', fontsize=12)
ax.set_ylabel('Quantidade de Clientes', fontsize=12)
ax.set_xticklabels(['Não Cancelou', 'Cancelou'], rotation=0)
ax.grid(axis='y', alpha=0.3)
for i, v in enumerate(cancelamento_counts.values):
    ax.text(i, v + 500, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Gráfico: Taxa de cancelamento por duração de contrato
fig, ax = plt.subplots(figsize=(10, 6))
taxa_cancelamento = tabela.groupby('duracao_contrato')['cancelou'].agg(['sum', 'count'])
taxa_cancelamento['taxa'] = (taxa_cancelamento['sum'] / taxa_cancelamento['count'] * 100)
taxa_cancelamento['taxa'].plot(kind='bar', ax=ax, color=['#3498db', '#e67e22', '#9b59b6'], edgecolor='black', linewidth=1.5)
ax.set_title('Taxa de Cancelamento por Duração do Contrato', fontsize=14, fontweight='bold')
ax.set_xlabel('Duração do Contrato', fontsize=12)
ax.set_ylabel('Taxa de Cancelamento (%)', fontsize=12)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45)
ax.grid(axis='y', alpha=0.3)
for i, v in enumerate(taxa_cancelamento['taxa'].values):
    ax.text(i, v + 1, f'{v:.1f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Gráfico: Cancelamentos por tipo de assinatura
fig, ax = plt.subplots(figsize=(10, 6))
taxa_assinatura = tabela.groupby('assinatura')['cancelou'].agg(['sum', 'count'])
taxa_assinatura['taxa'] = (taxa_assinatura['sum'] / taxa_assinatura['count'] * 100)
taxa_assinatura['taxa'].plot(kind='bar', ax=ax, color=['#e74c3c', '#f39c12', '#9b59b6'], edgecolor='black', linewidth=1.5)
ax.set_title('Taxa de Cancelamento por Tipo de Assinatura', fontsize=14, fontweight='bold')
ax.set_xlabel('Tipo de Assinatura', fontsize=12)
ax.set_ylabel('Taxa de Cancelamento (%)', fontsize=12)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45)
ax.grid(axis='y', alpha=0.3)
for i, v in enumerate(taxa_assinatura['taxa'].values):
    ax.text(i, v + 1, f'{v:.1f}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Gráfico: Influência de ligações do call center no cancelamento
fig, ax = plt.subplots(figsize=(12, 6))
taxa_callcenter = tabela.groupby('ligacoes_callcenter')['cancelou'].agg(['sum', 'count'])
taxa_callcenter['taxa'] = (taxa_callcenter['sum'] / taxa_callcenter['count'] * 100)
cores_gradient = plt.cm.RdYlGn_r(np.linspace(0.2, 0.8, len(taxa_callcenter)))
taxa_callcenter['taxa'].plot(kind='bar', ax=ax, color=cores_gradient, edgecolor='black', linewidth=1.5)
ax.set_title('Taxa de Cancelamento por Número de Ligações do Call Center', fontsize=14, fontweight='bold')
ax.set_xlabel('Número de Ligações do Call Center', fontsize=12)
ax.set_ylabel('Taxa de Cancelamento (%)', fontsize=12)
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
ax.grid(axis='y', alpha=0.3)
for i, v in enumerate(taxa_callcenter['taxa'].values):
    ax.text(i, v + 1.5, f'{v:.1f}%', ha='center', fontweight='bold', fontsize=9)
plt.tight_layout()
plt.show()

print("\n⚠️  INSIGHT IMPORTANTE: A taxa de cancelamento aumenta significativamente com o número de ligações do call center!")
print("Isso sugere que contatos frequentes podem estar associados a clientes insatisfeitos.")
```

```python
# Gráfico: Correlação entre dias de atraso e cancelamento
fig, ax = plt.subplots(figsize=(12, 6))
cancelados = tabela[tabela['cancelou'] == 1]['dias_atraso']
nao_cancelados = tabela[tabela['cancelou'] == 0]['dias_atraso']

bp = ax.boxplot([nao_cancelados, cancelados], labels=['Não Cancelou', 'Cancelou'], patch_artist=True)
colors = ['#2ecc71', '#e74c3c']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

ax.set_title('Distribuição de Dias de Atraso por Status de Cancelamento', fontsize=14, fontweight='bold')
ax.set_ylabel('Dias de Atraso', fontsize=12)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nMédia de dias de atraso - Não cancelaram: {nao_cancelados.mean():.2f} dias")
print(f"Média de dias de atraso - Cancelaram: {cancelados.mean():.2f} dias")
```

```python
# Gráfico: Relação entre total gasto e cancelamento
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

cancelados_gasto = tabela[tabela['cancelou'] == 1]['total_gasto']
nao_cancelados_gasto = tabela[tabela['cancelou'] == 0]['total_gasto']

axes[0].hist([nao_cancelados_gasto, cancelados_gasto], label=['Não Cancelou', 'Cancelou'], 
             color=['#2ecc71', '#e74c3c'], alpha=0.7, bins=30, edgecolor='black')
axes[0].set_title('Distribuição de Total Gasto', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Total Gasto', fontsize=11)
axes[0].set_ylabel('Frequência', fontsize=11)
axes[0].legend()
axes[0].grid(axis='y', alpha=0.3)

bp = axes[1].boxplot([nao_cancelados_gasto, cancelados_gasto], labels=['Não Cancelou', 'Cancelou'], patch_artist=True)
colors = ['#2ecc71', '#e74c3c']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
axes[1].set_title('Comparação de Total Gasto', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Total Gasto', fontsize=11)
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\nMédia de gasto - Não cancelaram: ${nao_cancelados_gasto.mean():.2f}")
print(f"Média de gasto - Cancelaram: ${cancelados_gasto.mean():.2f}")
```
